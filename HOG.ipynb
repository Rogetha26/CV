```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage.feature import hog

# Load image
image = cv2.imread('red_apple.jpg', cv2.IMREAD_GRAYSCALE)  #jpg


#img = cv2.imread('red_apple.png')
# Save it as JPG
#cv2.imwrite('image.jpg', img)


# Compute HOG
features, hog_image = hog(image, orientations=9, pixels_per_cell=(8, 8),
                          cells_per_block=(2, 2), visualize=True)

height, width = hog_image.shape
bright_image = np.zeros((height, width), dtype=np.uint8)
brightness = 20;
    
for i in range(height):
    for j in range(width):
        value = hog_image[i, j]*brightness
        bright_image[i, j] = value.clip(0, 255)

```

```python
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.imshow(image, cmap='gray')
plt.title('Original Image')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(bright_image, cmap='gray')
plt.title('HOG Image')
plt.axis('off')

plt.show()
```
